## Product Infomation


product_id: The unique identifier for the product from the site

product_name: The full name of the product

brand_id: The unique identifier for the product brand from the site

brand_name: The full name of the product brand

loves_count: The number of people who have marked this product as a favorite

rating: The average rating of the product based on user reviews

reviews: The number of user reviews for the product

size: The size of the product, which may be in oz, ml, g, packs, or other units depending on the product type

variation_type: The type of variation parameter for the product (e.g. Size, Color)

variation_value: The specific value of the variation parameter for the product (e.g. 100 mL, Golden Sand)

variation_desc: A description of the variation parameter for the product (e.g. tone for fairest skin)

ingredients: A list of ingredients included in the product, for example: [‘Product variation 1:’, ‘Water, Glycerin’, ‘Product variation 2:’, ‘Talc, Mica’] or if no variations [‘Water, Glycerin’]

price_usd: The price of the product in US dollars

value_price_usd: The potential cost savings of the product, presented on the site next to the regular price

sale_price_usd: The sale price of the product in US dollars

limited_edition: Indicates whether the product is a limited edition or not (1-true, 0-false)

new: Indicates whether the product is new or not (1-true, 0-false)

online_only: Indicates whether the product is only sold online or not (1-true, 0-false)

out_of_stock: Indicates whether the product is currently out of stock or not (1 if true, 0 if false)

sephora_exclusive: Indicates whether the product is exclusive to Sephora or not (1 if true, 0 if false)

highlights: A list of tags or features that highlight the product's attributes (e.g. [‘Vegan’, ‘Matte Finish’])

primary_category: First category in the breadcrumb section

secondary_category: Second category in the breadcrumb section

tertiary_category: Third category in the breadcrumb section

child_count: The number of variations of the product available

child_max_price: The highest price among the variations of the product

child_min_price: The lowest price among the variations of the product

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Analyzing categorical columns: Most common brands and product categories
top_brands = filtered_data["brand_name"].value_counts().head(10)
top_primary_categories = filtered_data["primary_category"].value_counts().head(5)
top_secondary_categories = filtered_data["secondary_category"].value_counts().head(10)
top_tertiary_categories = filtered_data["tertiary_category"].value_counts().head(10)

fig, axes = plt.subplots(2, 2, figsize=(15, 10))  # Create a 2x2 grid of subplots

# Top 10 Brands
sns.barplot(y=top_brands.index, x=top_brands.values, ax=axes[0,0])
axes[0,0].set_title("Top 10 Brands by Number of Products")
axes[0,0].set_xlabel("Number of Products")
axes[0,0].set_ylabel("Brand Name")

# Top 5 Primary Categories
sns.barplot(y=top_primary_categories.index, x=top_primary_categories.values, ax=axes[0,1])
axes[0,1].set_title("Top 5 Primary Categories by Number of Products")
axes[0,1].set_xlabel("Number of Products")
axes[0,1].set_ylabel("Primary Category")

# Top 10 Secondary Categories
sns.barplot(y=top_secondary_categories.index, x=top_secondary_categories.values, ax=axes[1,0])
axes[1,0].set_title("Top 10 Secondary Categories by Number of Products")
axes[1,0].set_xlabel("Number of Products")
axes[1,0].set_ylabel("Secondary Category")

# Top 10 Tertiary Categories
sns.barplot(y=top_tertiary_categories.index, x=top_tertiary_categories.values, ax=axes[1,1])
axes[1,1].set_title("Top 10 Tertiary Categories by Number of Products")
axes[1,1].set_xlabel("Number of Products")
axes[1,1].set_ylabel("Tertiary Category")

plt.tight_layout()
plt.show()
```

```python
# Distribution of Ratings
plt.figure(figsize=(8, 5))
sns.histplot(filtered_data["rating"], bins=10)
plt.title("Distribution of Ratings")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

```python
# Filter data for price range 0-500
filtered_price_data = filtered_data[(filtered_data["price"] >= 1) & (filtered_data["price"] <= 500)]  # Avoid zero values in log scale

# Calculate correlation
correlation = filtered_price_data[["price", "rating"]].corr().iloc[0, 1]

# Create scatter plot with regression line
plt.figure(figsize=(8, 5))
sns.regplot(
    x="price", 
    y="rating", 
    data=filtered_price_data, 
    scatter_kws={"alpha": 0.2},  # Adjust point transparency
    line_kws={"color": "red"}
)

# Apply log scale to x-axis
plt.xscale("log")

# Add correlation text
plt.text(10, 1.5, f"Correlation: {correlation:.2f}", fontsize=12, color="blue")

# Set title and labels
plt.title("Price vs. Rating (Filtered: $0 - $500, Log Scale)")
plt.xlabel("Price ($) [Log Scale]")
plt.ylabel("Rating")
plt.show()
```

```python
# Group by brand and calculate the average rating, then sort in descending order
top_rated_brands = filtered_data.groupby("brand_name")["rating"].mean().sort_values(ascending=False).head(10)

# Create bar plot
plt.figure(figsize=(10, 5))
sns.barplot(x=top_rated_brands.index, y=top_rated_brands.values)

# Adjust labels
plt.xticks(rotation=45)
plt.title("Top 10 Brands with Highest Average Rating")
plt.xlabel("Brand Name")
plt.ylabel("Average Rating")
plt.ylim(4, 5)  # Assuming ratings are between 1-5 for better visualization
plt.show()
```

```python
# Define categories to analyze
categories = ["skin_tone", "skin_type", "eye_color", "hair_color"]

# Create 2x2 subplot grid
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle("Distribution of Ratings by Skin Tone, Skin Type, Eye Color, and Hair Color", fontsize=16)

# Flatten the axes array for easier iteration
axes = axes.flatten()

# Plot each category in its respective subplot
for i, category in enumerate(categories):
    sns.barplot(x=category, y="rating", data=filtered_data, ax=axes[i])
    axes[i].set_xticklabels(axes[i].get_xticklabels(), rotation=45)
    axes[i].set_title(f"Ratings by {category.replace('_', ' ').title()}")
    axes[i].set_xlabel(category.replace("_", " ").title())
    axes[i].set_ylabel("Rating")
    axes[i].set_ylim(3, 5)  # Keep rating range consistent

# Adjust layout
plt.tight_layout(rect=[0, 0, 1, 0.96])  # Avoid overlap with suptitle
plt.show()
```
