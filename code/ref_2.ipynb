```python
# Simple example of using the text analysis functions
import pandas as pd
from fixed_text_analysis import initialize_nltk_resources, analyze_reviews, print_analysis_results, quick_review_analysis

# Step 1: Make sure to initialize NLTK resources first
# This will download necessary files if they don't exist
print("Initializing NLTK resources...")
initialize_nltk_resources()

# Step 2: Load your review data
# If you have a CSV file:
reviews_file = "../data/reviews_1250-end.csv"  # Replace with your file path

# Option A: Try the full analysis
try:
    print("\nRunning full analysis...")
    results, processed_df = analyze_reviews(reviews_file)
    print_analysis_results(results)
except Exception as e:
    print(f"Error in full analysis: {e}")
    print("Falling back to simplified analysis...")
    
    # Option B: Run simplified analysis that doesn't rely heavily on NLTK
    reviews_df = pd.read_csv(reviews_file)
    results = quick_review_analysis(reviews_df)
    print_analysis_results(results)

# Step 3: Extract specific insights (this will work with either analysis method)

# What do customers say most often in positive reviews?
print("\n----- WHAT CUSTOMERS LIKE MOST -----")
for term, count in results["positive_terms"][:10]:
    print(f"{term}: {count}")

# What do customers say most often in negative reviews?
print("\n----- WHAT CUSTOMERS DISLIKE MOST -----")
for term, count in results["negative_terms"][:10]:
    print(f"{term}: {count}")

# What skincare concerns do customers mention?
print("\n----- COMMON SKINCARE CONCERNS -----")
skincare_concerns = results["skincare_concerns"]
for concern, count in sorted(skincare_concerns.items(), key=lambda x: x[1], reverse=True):
    if count > 0:  # Only show concerns that were actually mentioned
        print(f"{concern}: {count}")
```
